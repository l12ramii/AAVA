```python
import sys
import platform

print('Python:', sys.version.split()[0])
print('Sistema:', platform.platform())
```

## 1. Comprobación de librerías básicas


```python
packages = {
    'numpy': 'numpy',
    'opencv': 'cv2',
    'scipy': 'scipy',
    'matplotlib': 'matplotlib',
    'scikit-image': 'skimage',
    'scikit-learn': 'sklearn',
    'Pillow': 'PIL',
    'torch': 'torch',
    'torchvision': 'torchvision',
    'tqdm': 'tqdm',
}

results = {}

for name, module_name in packages.items():
    try:
        module = __import__(module_name)
        version = getattr(module, '__version__', 'versión no disponible')
        results[name] = True
        print(f'[OK] {name:15s} {version}')
    except Exception as e:
        results[name] = False
        print(f'[--] {name:15s} NO DISPONIBLE -> {type(e).__name__}: {e}')
```

## 2. Test funcional de NumPy + OpenCV

Este test crea una imagen sintética en memoria, dibuja un rectángulo y la convierte a escala de grises. No abre ninguna ventana gráfica.


```python
import numpy as np
import cv2

img = np.zeros((100, 100, 3), dtype=np.uint8)
cv2.rectangle(img, (20, 20), (80, 80), (255, 255, 255), thickness=-1)
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

print('Shape imagen:', img.shape)
print('Shape gris:', gray.shape)
print('Valor máximo:', gray.max())

assert img.shape == (100, 100, 3)
assert gray.shape == (100, 100)
assert gray.max() == 255

print('OK: NumPy y OpenCV funcionan correctamente.')
```

## 3. Comprobación de PyTorch y CUDA

En Google Colab, una GPU solo estará disponible si se ha seleccionado un entorno de ejecución con acelerador GPU.


```python
try:
    import torch
    print('PyTorch:', torch.__version__)
    print('CUDA disponible:', torch.cuda.is_available())

    if torch.cuda.is_available():
        print('GPU:', torch.cuda.get_device_name(0))
        print('Versión CUDA usada por PyTorch:', torch.version.cuda)
        print('Número de GPUs visibles:', torch.cuda.device_count())
    else:
        print('AVISO: PyTorch está funcionando sin GPU/CUDA.')
except Exception as e:
    print('PyTorch no está disponible o no se pudo inicializar.')
    print(type(e).__name__ + ':', e)
```

## 4. Información opcional de la GPU del sistema

En Colab con GPU, `nvidia-smi` permite comprobar el dispositivo y el controlador instalados.


```python
import shutil
import subprocess

if shutil.which('nvidia-smi'):
    result = subprocess.run(
        ['nvidia-smi'],
        capture_output=True,
        text=True
    )
    print(result.stdout)
else:
    print('nvidia-smi no está disponible en este entorno.')
```

## 5. Resumen


```python
print('Resumen del entorno')
print('-' * 40)
for name, ok in results.items():
    print(f'{name:15s}: ' + ('OK' if ok else 'NO DISPONIBLE'))

missing = [name for name, ok in results.items() if not ok]

if not missing:
    print('\nEntorno básico preparado correctamente.')
else:
    print('\nFaltan estas librerías:')
    for name in missing:
        print(' -', name)
```

```python
# Ejercicio 1
import cv2

img = cv2.imread("./images/cells.png")
if img is None: raise FileNotFoundError("No se pudo cargar la imagen")

print("Tipo:", type(img)) 
print("Dimensiones:", img.shape) 
print("Tipo de dato:", img.dtype) 
print("Valor mínimo:", img.min()) 
print("Valor máximo:", img.max())
```

```python
# Ejercicio 2

height, width = img.shape[:2]

x, y = height // 2, width // 2
pixel = img[x,y]
print(pixel)

y1, y2 = height // 4, 3* height // 4
x1, x2 = width //4, 3 * width //4
modified = img.copy()
modified[y1:y2, x1:x2] = 0
cv2.imwrite("modified.png", modified) 

roi = img[y1:y2, x1:x2]
cv2.imwrite("roi.png", roi)

roi_view = img[y1:y2, x1:x2]
roi_copy = roi_view.copy()
roi_view[:] = 0
roi_copy[:] = 255
```

```python
# Ejercicio 3
import matplotlib.pyplot as plt

img = cv2.imread("./images/dogmask.jpg")
b, g, r = cv2.split(img)
cv2.imwrite("canal_b.png", b)
cv2.imwrite("canal_g.png", g)
cv2.imwrite("canal_r.png", r)

rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
plt.imshow(rgb)
plt.axis("off")
plt.show();

gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
plt.imshow(gray, cmap="gray")
plt.axis("off")
plt.show();

print(f"dimensiones bgr: {img.shape}")
print(f"dimensiones gris: {gray.shape}")
```

```python
# Ejercicio 4
small = cv2.resize(img, None, fx = 0.5, fy = 0.5, interpolation = cv2.INTER_AREA)
cv2.imwrite("small.png", small)

horizontal = cv2.flip(img, 1)
vertical = cv2.flip(img, 0)
cv2.imwrite("flip_horizontal.png", horizontal) 
cv2.imwrite("flip_vertical.png", vertical)

rotated = cv2.rotate(img, cv2.ROTATE_90_CLOCKWISE)
cv2.imwrite("rotated.png", rotated)
```

```python
# Ejercicio 5
blur3x3 = cv2.GaussianBlur(img,(3,3),0)
cv2.imwrite("blur3x3.png", blur3x3)
blur5x5 = cv2.GaussianBlur(img,(5,5),0)
cv2.imwrite("blur5x5.png", blur5x5)
blur11x11 = cv2.GaussianBlur(img,(11,11),0)
cv2.imwrite("blur11x11.png", blur11x11)
blur21x21 = cv2.GaussianBlur(img,(21,21),0)
cv2.imwrite("blur21x21.png", blur21x21)

img = cv2.imread("./images/cells.png")
img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
edges = cv2.Canny(img_gray,100,200)
cv2.imwrite("edges.png", edges)

img_gray = cv2.GaussianBlur(img_gray, (5,5),0)
edges = cv2.Canny(img_gray,100,200)
cv2.imwrite("edges2.png", edges)

```

```python
# Ejercicio 6

_, mask = cv2.threshold(gray, 128,255,cv2.THRESH_BINARY)
cv2.imwrite("mask_threshold.png", mask)

threshold, mask_otsu= cv2.threshold(gray, 0, 255, cv2.THRESH_OTSU)
print("Umbral de Otsu:", threshold) 
cv2.imwrite("mask_otsu.png", mask_otsu)
```
